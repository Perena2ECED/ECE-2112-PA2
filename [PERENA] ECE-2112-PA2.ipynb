# A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
```

```python
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
mean = np.mean(X)
mean
```

```python
std = np.std(X)
std
```

```python
X_normalized = (X-mean)/std
X_normalized
```

```python
m = np.mean(X_normalized)
m
```

```python
s = np.std(X_normalized)
s
```

```python
np.save("X_normalized.npy", X_normalized)
```

# B. CUBES DIVISIBLE BY 4

```python
C = np.arange(1, 101)
C
```

```python
C = C**3
C
```

```python
C = C.reshape(10, 10)
C
```

```python
C.shape
```

```python
div_by_4 = C[C%4==0]
div_by_4
```

```python
P1 = div_by_4.size
P1
```

```python
np.save("div_by_4.npy", div_by_4)
```

# C. ABOVE-MEAN SQUARES PROBLEMS

```python
S = np.arange(1, 37)
S
```

```python
S = np.power(S, 2)
S
```

```python
S = S.reshape(6, 6)
S
```

```python
S.shape
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S>S_mean]
above_mean
```

```python
P3 = len(above_mean)
P3
```

```python
np.save("above_mean.npy", above_mean)
```
